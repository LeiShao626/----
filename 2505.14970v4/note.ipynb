### 人工设计的课程所谓启发式规则
    也就是基于人的主观经验、常识或预设的固定逻辑，而非基于模型实际学习状态来决定训练顺序的方法。
    

### 在线过滤方法（Online Filtering Methods）
    就是选出本轮训练中最适合模型当前学习水平的那些样本，太容易的样本，模型获得的奖励满分，不要，太难的，零分，也不要。举个例子，对于一个初中生，老师给了他一堆题目，有小学题，有初中题，有高中题，小学题全对，初中题半对半错，高中题全错，很显然，此时最能提升他水平的就是初中题，于是过滤了小学题和高中题。
    在线过滤方法很大的问题在于计算量巨大，因为为了筛选出初中题，你不得不耗费算力让模型去回答那些小学题和高中题。

### 在线（Online）定义
    模型训练是一个与环境实时交互的闭环过程。策略（模型）在当前状态（题目）下生成动作（推理轨迹），立刻获得环境的奖励（评分）反馈，并使用当前轨迹产生的梯度即时更新参数。依赖当前策略自身的采样，注重实时性与探索-利用权衡。

### 离线学习（Offline Learning / Off-policy）
    模型只看别人做过的旧题和旧答案（静态数据集），照着背，学完后才去考试。核心是“拿别人或过去的旧数据，一次性学到底”，训练时不与环境交互。
